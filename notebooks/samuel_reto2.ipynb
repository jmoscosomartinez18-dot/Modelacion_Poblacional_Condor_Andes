```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Leer datos
df = pd.read_csv('../data/datos_poblacion_condor.csv')

# Mostrar datos
print("DATOS ORIGINALES")
print(df)

# Limpiar datos
df.dropna(inplace=True)

# Convertir tipos
df['Año'] = df['Año'].astype(int)
df['Poblacion'] = df['Poblacion'].astype(float)

print("\nDATOS LIMPIOS")
print(df)

# Calcular población inicial
P0 = df['Poblacion'].iloc[0]

print("\nPoblación inicial P0:")
print(P0)

# Estimar tasa de crecimiento promedio

años = df['Año']
poblacion = df['Poblacion']

r = np.mean(
    np.diff(np.log(poblacion)) /
    np.diff(años)
)

print("\nTasa de crecimiento estimada r:")
print(r)

# Graficar

plt.figure(figsize=(8,5))

plt.plot(
    df['Año'],
    df['Poblacion'],
    marker='o'
)

plt.title("Población del Cóndor de los Andes")

plt.xlabel("Año")
plt.ylabel("Número de individuos")

plt.grid()

plt.show()
```
