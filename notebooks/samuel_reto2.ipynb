```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Leer CSV
df = pd.read_csv('../data/datos_poblacion_condor.csv')

# Mostrar nombres reales de columnas
print("COLUMNAS ENCONTRADAS:")
print(df.columns.tolist())

# Renombrar automáticamente las columnas
df.columns = ['Año','Poblacion']

# Mostrar datos
print("\nDATOS:")
print(df)

# Limpiar datos
df.dropna(inplace=True)

# Convertir tipos
df['Año'] = pd.to_numeric(df['Año'])
df['Poblacion'] = pd.to_numeric(df['Poblacion'])

# Población inicial
P0 = df['Poblacion'].iloc[0]

print("\nPoblación inicial:")
print(P0)

# Calcular tasa r
años = df['Año']
poblacion = df['Poblacion']

r = np.mean(
    np.diff(np.log(poblacion))
    /
    np.diff(años)
)

print("\nTasa de crecimiento r:")
print(r)

# Graficar
plt.figure(figsize=(10,5))

plt.plot(
    años,
    poblacion,
    marker='o'
)

plt.title("Población del Cóndor de los Andes")

plt.xlabel("Año")
plt.ylabel("Población")

plt.grid(True)

plt.show()
```
